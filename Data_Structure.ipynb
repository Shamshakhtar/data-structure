Qno. 2.  Explain the difference between mutable and immutable data types with examples.

Answer - Mutable - A mutable data type can be changed or modified after its creation, allowing you to update, add, or remove elements.


```python
# Example: list, dict, set
lis = [1, 2, 3]
lis.append(4)
lis                          #output: [1,2,3,4]
```

Immutable - An immutable data type cannot be changed after its creation; if you try to modify it, a new object is created instead.

```python
# Example: tuple
t = (1, 2, 3)
t[1] = 4
```

Qno. 3.  What are the main differences between lists and tuples in Python?

Answer - Lists are mutable, which means their contents can be modified, while tuples are immutable and remain unchanged after creation. Lists are ideal for handling dynamic data, whereas tuples are best suited for storing fixed or constant information.

Qno. 5. Why might you use a set instead of a list in Python?

Qno. 8.  What is a hash table, and how does it relate to dictionaries in Python

Answer - A hash table is a fast and efficient data structure used to store key-value pairs. In Python, a dict is implemented using a hash table, where each key-value pair is stored based on the hash of the key, making key-based lookups extremely fast.

```python
#example:
my_dict = {"name": "Alice","age": 30}

print(my_dict["name"])
```

Qno. 14- How does the “in” keyword work differently for lists and dictionaries

Answer - The in keyword is used to check for membership — but it behaves differently depending on the data type:


```python
# list - in checks if a value exists as an element in the list.
my_list = [1, 2, 3, 4]
print(2 in my_list)
print(5 in my_list)
```

```python
# Dict - in checks if a value exists as a key in the dictionary — not the value by default.
my_dict = {"name": "shamsh", "age": 25}
print("Shamsh" in my_dict)
print("25" in my_dict)
```

Qno.16 - What is a nested dictionary, and give an example of its use case?

Answer - A nested dictionary is a dictionary inside another dictionary. It allows you to store complex, structured data in a neat and organized way using multiple levels.




```python
#Example:
timetable = {"Monday": {"9AM": "Math","10AM": "Science"},
    "Tuesday": {"9AM": "English","10AM": "History" }}
print(timetable)

```

 Qno. 17. Describe the time complexity of accessing elements in a dictionary?

 Answer - Accessing elements in a dictionary is O(1) on average due to its hash table structure, making it much faster than searching through lists (O(n)).


```python
#example
student = {"name": "Alice", "age": 20}
(student["age"])  # O(1) time to access "age"
```

Qno. 18. -  In what situations are lists preferred over dictionaries.

Answer - Use lists when you're working with simple, ordered collections of values and don't need to associate each item with a specific key. They're great for sequences, queues, stacks, or any indexed data.

Qno. 19-  Why are dictionaries considered unordered, and how does that affect data retrieval?

Answer - Dictionaries are considered unordered because their primary goal is efficient key-based access, not maintaining order. While Python 3.7+ preserves insertion order, dictionaries are still designed for fast lookups, not position-based access like lists.




```python
#eaxmple:
data = {"name": "Alice", "age": 30, "city": "London"}
print(data["age"])
```
